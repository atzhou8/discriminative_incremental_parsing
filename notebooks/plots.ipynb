```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv('metrics/items_ClassicGP.parser.csv')
```

```python
amb = data[data['ambiguity'] == 'ambiguous']
unamb = data[data['ambiguity'] == 'unambiguous']


def sum_window(df, disamb_col, condition_col):
    return (
        df.assign(
            in_window=lambda x: (x['word_pos'] >= x[disamb_col]) &
                                 (x['word_pos'] <= x[disamb_col] + 2),
            condition=lambda x: x[condition_col].str.replace(r'_(AMB|UAMB)$', '', regex=True),
        )
        .loc[lambda x: x['in_window'], ['item', 'condition', 'kl_backward']]
        .groupby(['item', 'condition'], as_index=False)['kl_backward']
        .sum()
        .rename(columns={'kl_backward': 'kl_backward_sum'})
    )


amb_cp = sum_window(amb, 'disambPositionAmb', 'condition')
unamb_cp = sum_window(unamb, 'disambPositionUnamb', 'condition')

merged = amb_cp.merge(unamb_cp, on=['item', 'condition'], suffixes=('_amb', '_unamb'))
merged['GPE'] = merged['kl_backward_sum_amb'] - merged['kl_backward_sum_unamb']

print(merged.groupby('condition', as_index=False)['GPE'].mean())
```
